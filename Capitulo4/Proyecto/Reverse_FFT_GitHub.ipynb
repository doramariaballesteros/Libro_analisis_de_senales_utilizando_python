**Transformada de Fourier de Señales periódicas y aperiódicas**

Este notebook acompaña el proyecto de **Análisis de Señales** y proporciona el código para recuperar mensajes secretos previamente codificados mediante la permutación de coeficientes espectrales complejos.

El procedimiento utiliza la **Transformada Rápida de Fourier para señales reales (`rfft`)** y su transformada inversa (`irfft`), junto con una clave de permutación que permite reconstruir el mensaje secreto original.





### 🎯 Objetivo

Recuperar el mensaje secreto original a partir del audio codificado (**Encoded**) y su correspondiente clave de permutación.

### 🔄 Procedimiento

1. 📂 **Carga** el audio codificado (`encoded.wav`) y la clave de permutación (`FFT_key.csv`).
2. 📊 **Calcula** los coeficientes espectrales complejos del audio codificado utilizando `rfft`.
3. 🔑 **Invierte** la permutación mediante la clave, recuperando las posiciones originales de los coeficientes complejos.
4. 🎵 **Aplica** `irfft` para reconstruir la señal secreta en el dominio temporal.
5. 💾 **Guarda** el mensaje recuperado como `recovered.wav` y escúchalo para verificar su inteligibilidad.

### ✅ Resultado esperado

El audio recuperado debe contener el mensaje secreto original y permitir su reconocimiento.

📌 **Importante:** la recuperación utiliza los coeficientes complejos completos, conservando tanto su parte real como su parte imaginaria. No es necesario almacenar ni reconstruir la fase por separado.


```python
import numpy as np
from scipy.io.wavfile import read, write
from scipy.fft import rfft, irfft
from IPython.display import Audio, display

# 1. CARGAR AUDIO CODIFICADO Y CLAVE
fs, encoded = read('encoded.wav')
key = np.loadtxt('FFT_key.csv', delimiter=',', skiprows=1, dtype=int)

# 2. TRANSFORMADA DE FOURIER
X_encoded = rfft(encoded.astype(float))

# 3. DESPERMUTAR COEFICIENTES COMPLEJOS
X_recovered = np.empty_like(X_encoded)
X_recovered[key] = X_encoded

# 4. TRANSFORMADA INVERSA
recovered = irfft(X_recovered, n=len(encoded))

# 5. GUARDAR Y ESCUCHAR
write('recovered.wav', fs, recovered.astype(np.float32))
display(Audio(recovered, rate=fs))
```
